```python
import cv2
import numpy as np
import dlib
import face_recognition as fr

print("OpenCV:", cv2.__version__)
print("dlib:", dlib.__version__)
print("face_recognition:", fr.__version__)
```

```python
video_capture=cv2.VideoCapture(0)
```

```python
image = fr.load_image_file(r'C:\Users\Admin\Pictures\Camera Roll\WIN_20260901_12_39_01_Pro.jpg')
```

```python
image_face_encoding=fr.face_encodings(image)[0]
```

```python
known_face_encodings=[image_face_encoding]
```

```python
known_face_names = ["ishi"]
```

```python
while True:
    ret,frame=video_capture.read()
    
    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    
    fc_locations=fr.face_locations(rgb_frame)
    face_encodings=fr.face_encodings(rgb_frame,fc_locations)
    
    for(top,right,bottom,left),face_encoding in zip(fc_locations,face_encodings):
        
        matches=fr.compare_faces(known_face_encodings,face_encoding)
        name="unknown"
        
        fc_distances=fr.face_distance(known_face_encodings,face_encoding)
        match_index = np.argmin(fc_distances)
       
        if matches[match_index]:
            name = known_face_names[match_index]
        
        cv2.rectangle(frame, (left, top), (right, bottom), (0,0,255), 2)
        
        cv2.rectangle(frame, (left, bottom - 35), (right, bottom), (0,0,255), cv2.FILLED)
        
        font = cv2.FONT_HERSHEY_SIMPLEX
        cv2.putText(frame,name,(left+6,bottom-6),font,1.0,(255,255,255),1)
    
    cv2.imshow('Simplelearn face detection system', frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

video_capture.release()
cv2.destroyAllWindows()
```
